### Pyspark Datypes vs Data Frames

- PySpark DataFrames are designed for large datasets distributed across multiple machines, unlike Pandas DataFrames which work on a single machine's memory.
- PySpark DataFrames are immutable, so operations create new DataFrames, ensuring consistency in distributed environments.
- PySpark uses lazy evaluation, building a directed acyclic graph (DAG) of operations that Spark optimizes and executes only when an action is triggered, improving efficiency.

### Creating New Columns

- You can create new columns in a DataFrame using the withColumn method, such as calculating trip duration by subtracting pickup from dropoff times and converting the result to minutes.
- Renaming columns for consistency is easy with the withColumnsRenamed method, which takes a dictionary mapping old names to new names.
- To focus on relevant data, use the select statement to pick specific columns and the drop method to remove unwanted columns from your DataFrame.

```python
from pyspark.sql.functions import unix_timestamp, round
```

```python
df1 = df.withColumn(
    'trip_duration', 
    round((unix_timestamp("tpep_dropoff_datetime") - unix_timestamp("tpep_pickup_datetime") / 60),1)
)
```

```python
# to rename columns using .withColumnRenamed()
df2 = df1.select(
    'tpep_pickup_datetime',
    'tpep_dropoff_datetime',
    'trip_duration',
    'PULocationID',
    'DOLocationID',
    'passenger_count',
    'fare_amount',
    'Airport_fee',
    'total_amount'
) \
.withColumnRenamed('tpep_pickup_datetime', 'pu_datetime') \
.withColumnRenamed('tpep_dropoff_datetime', 'do_datetime') \
.withColumnRenamed('PULocationID', 'pu_location_id') \
.withColumnRenamed('DOLocationID', 'do_location_id') \
.withColumnRenamed('Airport_fee', 'airport_fee')
```

```python
df2.show()
```

```python
# cleaner way to rename columns using alias
from pyspark.sql.functions import col

df2 = df1.select(
    col('tpep_pickup_datetime').alias('pu_datetime'),
    col('tpep_dropoff_datetime').alias('do_datetime'),
    col('trip_duration').alias('trip_duration_minutes'),
    col('PULocationID').alias('pu_location_id'),
    col('DOLocationID').alias('do_location_id'),
    'passenger_count',
    'fare_amount',
    col('Airport_fee').alias('airport_fee'),
    'total_amount'
)
```

```python
df2.show()
```

```python
# To drop columns
df1.drop('vendor_id','ratecode_id').show()
```

### Creating Temporary views

- To use PySpark SQL, you first create a temporary view from your DataFrame, which registers the data with Spark SQL without copying it.
- The temporary view allows you to run SQL queries on the data within the current Spark session.
-Temporary views do not persist beyond the current session, so they need to be recreated if the session restarts

```python
taxi = spark.read.parquet("C:/Users/sahil/Desktop/Projects/Learning/Pyspark/Data/yellow_tripdata_2025-01.parquet")
```

```python
taxi.createOrReplaceTempView("taxi")
```

- PySpark SQL allows you to write SQL queries directly within your Python code, giving you flexibility in how you manipulate and analyze data.
- You can combine SQL queries with PySpark DataFrame API methods in the same workflow to fine-tune your data processing.
- Using the Spark session's SQL method returns a DataFrame, so you can chain further DataFrame operations like filtering and selecting columns.
- For readability and easier maintenance, especially with complex queries, it's helpful to assign SQL queries to a separate string variable using Python's triple-quoted string syntax.

### Challenge 3

- Load the January taxi ride data and the taxi zone lookup data into DataFrames and register them as temporary views.
- Use PySpark SQL to perform a left join on these two datasets based on location IDs.
- Select the drop-off location ID, borough, and total amount columns.
- Then, group the data by borough and calculate the average total ride cost.

```python
taxi = spark.read.parquet("C:/Users/sahil/Desktop/Projects/Learning/Pyspark/Data/yellow_tripdata_2025-01.parquet")
taxi.createOrReplaceTempView("taxi")
```

```python
Taxi_zone = spark.read.option("header", "true").csv("C:/Users/sahil/Desktop/Projects/Learning/Pyspark/Data/taxi_zone_lookup.csv")
Taxi_zone.createOrReplaceTempView("Taxi_Zone")
```

```python
spark.catalog.listTables()
```

```python
Query = '''
            SELECT 
                
                t.DOLocationID,
                tz.Borough,
                avg(t.total_amount) AS avg_total_ride_cost
            FROM 
                taxi as t
            LEFT JOIN
                Taxi_zone as tz
            ON  
                t.DOlocationID = tz.LocationID 
            GROUP BY
                t.DOLocationID,
                tz.Borough
            ORDER BY
                avg_total_ride_cost DESC
        '''
spark.sql(Query).show()
```

### Pyspark in Production Environment

- Production pipelines handle large-scale data across multiple servers or cloud instances, requiring scalability, reliability, and security.
